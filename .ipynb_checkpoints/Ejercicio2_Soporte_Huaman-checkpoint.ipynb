## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Tiempo de espera del ticket, nivel de urgencia y condición de cliente premium |
| Acciones (A) | Mantener en nivel 1, escalar a nivel 2 o escalar a nivel 3 |
| Entorno (E) | Mesa de ayuda, clientes, técnicos de soporte y cola de tickets |
| Objetivo | Asignar cada ticket al nivel adecuado según su prioridad y tiempo de espera |
| Medida de desempeño (P) | Reducir tiempos de atención, priorizar correctamente los casos urgentes y evitar escalaciones innecesarias |

### Justificación de las reglas

Las reglas serán las siguientes:

- Si la urgencia es alta, el ticket se escala inmediatamente a nivel 3.
- Si la urgencia es media y el cliente es premium, se escala a nivel 2.
- Si la urgencia es media y el tiempo de espera es de 30 minutos o más, se escala a nivel 2.
- Si la urgencia es baja pero el tiempo de espera es de 60 minutos o más, se escala a nivel 2.
- Si la urgencia es baja, el cliente es premium y ha esperado 40 minutos o más, se escala a nivel 2.
- En los demás casos, el ticket se mantiene en nivel 1.

Estas reglas permiten atender primero los casos más críticos, pero también consideran el tiempo acumulado de espera y la prioridad de los clientes premium. De esta forma se busca reducir retrasos excesivos sin escalar innecesariamente todos los tickets.

### Código

def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    urgencia = nivel_urgencia.lower()

    if urgencia == "alta":
        return (
            "escalar a nivel 3",
            "urgencia alta reportada por el cliente"
        )

    elif urgencia == "media":
        if cliente_premium:
            return (
                "escalar a nivel 2",
                "urgencia media y cliente premium"
            )

        elif tiempo_espera_minutos >= 30:
            return (
                "escalar a nivel 2",
                "urgencia media con tiempo de espera elevado"
            )

        else:
            return (
                "mantener en nivel 1",
                "urgencia media con tiempo de espera reducido"
            )

    elif urgencia == "baja":
        if cliente_premium and tiempo_espera_minutos >= 40:
            return (
                "escalar a nivel 2",
                "cliente premium con tiempo de espera elevado"
            )

        elif tiempo_espera_minutos >= 60:
            return (
                "escalar a nivel 2",
                "ticket de baja urgencia con tiempo de espera excesivo"
            )

        else:
            return (
                "mantener en nivel 1",
                "ticket de baja urgencia dentro del tiempo normal"
            )

    else:
        return (
            "error",
            "nivel de urgencia no válido"
        )

### Simulación y pruebas

```python
pruebas = [
    (5, "alta", False),
    (10, "media", False),
    (35, "media", False),
    (5, "media", True),
    (70, "baja", False),
    (45, "baja", True),
    (10, "baja", True)
]

for tiempo, urgencia, premium in pruebas:
    resultado = agente_soporte(
        tiempo,
        urgencia,
        premium
    )

    print(
        f"Tiempo de espera: {tiempo} min | "
        f"Urgencia: {urgencia} | "
        f"Cliente premium: {premium}"
    )

    print("Resultado:", resultado)
    print("-" * 60)
```

### Interpretación de los resultados

Las pruebas muestran que el agente prioriza los tickets de alta urgencia, mientras que los casos de urgencia media dependen del tiempo de espera y de la condición premium del cliente. Los tickets de baja urgencia se mantienen normalmente en nivel 1, salvo cuando el tiempo de espera es elevado o el cliente premium ha esperado demasiado.
