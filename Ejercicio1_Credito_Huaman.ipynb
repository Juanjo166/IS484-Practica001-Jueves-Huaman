## Ejercicio 1: Evaluación de solicitud de crédito bancario

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Ingreso mensual, monto solicitado e historial moroso del solicitante |
| Acciones (A) | Aprobar, aprobar con condiciones o rechazar la solicitud |
| Entorno (E) | Sistema de evaluación crediticia de una entidad bancaria |
| Objetivo | Evaluar el riesgo crediticio y determinar si la solicitud puede ser aprobada |
| Medida de desempeño (P) | Reducir créditos de alto riesgo y aprobar solicitudes con capacidad adecuada de pago |
