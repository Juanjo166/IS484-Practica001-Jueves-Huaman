## Ejercicio 1: Evaluación de solicitud de crédito bancario

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Ingreso mensual, monto solicitado e historial moroso del solicitante |
| Acciones (A) | Aprobar, aprobar con condiciones o rechazar la solicitud |
| Entorno (E) | Sistema de evaluación crediticia de una entidad bancaria |
| Objetivo | Evaluar el riesgo crediticio y determinar si la solicitud puede ser aprobada |
| Medida de desempeño (P) | Reducir créditos de alto riesgo y aprobar solicitudes con capacidad adecuada de pago |

### Justificación de las reglas

Para evaluar una solicitud se calculará la relación entre el monto solicitado y el ingreso mensual.

Las reglas serán las siguientes:

- Si no tiene historial moroso y la relación es menor o igual a 0.30, se aprueba.
- Si no tiene historial moroso y la relación está entre 0.30 y 0.50, se aprueba con condiciones.
- Si tiene historial moroso pero la relación es menor o igual a 0.30, se aprueba con condiciones.
- En los demás casos, se rechaza la solicitud.

Estas reglas consideran tanto la capacidad de pago como el historial del solicitante. Una relación monto/ingreso baja representa un menor riesgo para el banco. Cuando existe historial moroso se toma una decisión más conservadora. Las solicitudes con una relación elevada presentan mayor riesgo de incumplimiento.

### Código

```python
def agente_credito(ingreso_mensual, monto_solicitado, tiene_historial_moroso):
    relacion = monto_solicitado / ingreso_mensual

    if not tiene_historial_moroso and relacion <= 0.30:
        return "aprobar", f"relación monto/ingreso = {relacion:.2f} y sin historial moroso"

    elif not tiene_historial_moroso and relacion <= 0.50:
        return "aprobar con condiciones", f"relación monto/ingreso = {relacion:.2f} y sin historial moroso"

    elif tiene_historial_moroso and relacion <= 0.30:
        return "aprobar con condiciones", f"relación monto/ingreso = {relacion:.2f} pero presenta historial moroso"

    else:
        return "rechazar", f"relación monto/ingreso = {relacion:.2f} y nivel de riesgo alto"
```

### Simulación y pruebas

```python
pruebas = [
    (5000, 1000, False),  # relación 0.20
    (5000, 1500, False),  # relación 0.30 - caso límite
    (5000, 2000, False),  # relación 0.40
    (5000, 2500, False),  # relación 0.50 - caso límite
    (5000, 3000, False),  # relación 0.60
    (5000, 1000, True),   # moroso con relación baja
    (5000, 2000, True)    # moroso con relación media
]

for ingreso, monto, moroso in pruebas:
    relacion = monto / ingreso
    resultado = agente_credito(ingreso, monto, moroso)

    print(
        f"Ingreso: S/{ingreso} | "
        f"Monto: S/{monto} | "
        f"Relación: {relacion:.2f} | "
        f"Moroso: {moroso}"
    )

    print("Resultado:", resultado)
    print("-" * 60)
```
