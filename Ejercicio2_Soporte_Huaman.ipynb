## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Tiempo de espera del ticket, nivel de urgencia y condición de cliente premium |
| Acciones (A) | Mantener en nivel 1, escalar a nivel 2 o escalar a nivel 3 |
| Entorno (E) | Mesa de ayuda, clientes, técnicos de soporte y cola de tickets |
| Objetivo | Asignar cada ticket al nivel adecuado según su prioridad y tiempo de espera |
| Medida de desempeño (P) | Reducir tiempos de atención, priorizar correctamente los casos urgentes y evitar escalaciones innecesarias |

### Justificación de las reglas

Las reglas serán las siguientes:

- Si la urgencia es alta, el ticket se escala inmediatamente a nivel 3.
- Si la urgencia es media y el cliente es premium, se escala a nivel 2.
- Si la urgencia es media y el tiempo de espera es de 30 minutos o más, se escala a nivel 2.
- Si la urgencia es baja pero el tiempo de espera es de 60 minutos o más, se escala a nivel 2.
- Si la urgencia es baja, el cliente es premium y ha esperado 40 minutos o más, se escala a nivel 2.
- En los demás casos, el ticket se mantiene en nivel 1.

Estas reglas permiten atender primero los casos más críticos, pero también consideran el tiempo acumulado de espera y la prioridad de los clientes premium. De esta forma se busca reducir retrasos excesivos sin escalar innecesariamente todos los tickets.
