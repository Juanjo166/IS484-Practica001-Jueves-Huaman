## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Tiempo de espera del ticket, nivel de urgencia y condición de cliente premium |
| Acciones (A) | Mantener en nivel 1, escalar a nivel 2 o escalar a nivel 3 |
| Entorno (E) | Mesa de ayuda, clientes, técnicos de soporte y cola de tickets |
| Objetivo | Asignar cada ticket al nivel adecuado según su prioridad y tiempo de espera |
| Medida de desempeño (P) | Reducir tiempos de atención, priorizar correctamente los casos urgentes y evitar escalaciones innecesarias |
